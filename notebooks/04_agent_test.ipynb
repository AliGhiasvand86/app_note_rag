```python
# Test note creation

result = graph.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Create a note titled 'Agent Test' with the content 'Hello. This note was created by LangGraph.'"
            }
        ]
    },
    config={
        "configurable": {
            "user_id": 1
        }
    },
)

print(result["messages"][-1].content)
```

```python
# Test note summarization

result = graph.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Summarize the note titled 'Agent Test'."
            }
        ]
    },
    config={
        "configurable": {
            "user_id": 1
        }
    },
)

print(result["messages"][-1].content)
```

```python
# Test note deletion

result = graph.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Delete the note titled 'Agent Test'."
            }
        ]
    },
    config={
        "configurable": {
            "user_id": 1
        }
    },
)

print(result["messages"][-1].content)
```
