```python
# Import retrieval and reranking components

from src.rag.retrieval.hybrid import hybrid_search
from src.rag.retrieval.reranker import reranker
```

```python
# Initialize the database

from src.database import init_database

init_database()
```

```python
# Retrieve candidate chunks

query = "artificial intelligence research"

results = hybrid_search(
    query=query,
    user_id=4,
    top_k=10,
)

print(f"Retrieved candidates: {len(results)}")

for result in results:
    print(
        result["title"],
        "| RRF:",
        round(result["rrf_score"], 4),
    )
```

```python
# Rerank retrieved candidates

from src.rag.retrieval.reranker import reranker

reranked_results = reranker.rerank(
    query=query,
    results=results,
    top_k=3,
)

for result in reranked_results:
    print(
        result["title"],
        "| Reranker:",
        round(result["reranker_score"], 4),
    )
```
